# Feature Engineering - OneHot Encoding

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_csv("../data_set/Data_01.csv")
data
```

```python
le = LabelEncoder()
data["class"] = le.fit_transform(data.Purchased)
data
```

```python
data = data.drop(columns="Purchased")
data
```

```python
# categorical data
cat_cols = ["Country", "Sex"]
```

## `pd.get_dummies()` is a Pandas function that converts categorical variables into numerical (binary) columns. This process is called One-Hot Encoding.

```python
df = pd.get_dummies(data, columns=cat_cols)
df
```

```python
df = df.drop(columns=["Country_Bangladesh", "Sex_Female"])
df
```

```python
X = df.drop("class", axis=1)  # Feature Matrix
X
```
