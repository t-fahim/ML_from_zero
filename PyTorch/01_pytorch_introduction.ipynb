# PyTorch Introduction

```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
```

```python
print("pytorch version :", torch.__version__)
```

```python
# check gpu availability
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device :", device)
if torch.cuda.is_available():
    print("gpu name: ", torch.cuda.get_device_name(0))
```

## creating tensors

```python
# tensor form list
x = torch.tensor([1, 2, 3, 4])
print("x:", x)
```

```python
# float tensor
y = torch.tensor([1.0, 2.4, 5.0])
print("y:", y)
```

```python
# matrix
matrix = torch.tensor([[1, 2, 3], [4, 5, 6]])
print("martix:")
print(matrix)
```

## Tensor attribute

```python
x = torch.rand(3, 5)
print("tensor:")
print(x)
print("shape:", x.shape)
print("data type:", x.dtype)
print("device:", x.device)
```

## special tensor creation

```python
zeros = torch.zeros(3, 4)
ones = torch.ones(2, 5)
random_uniform = torch.rand(3, 3)
random_normal = torch.randn(3, 3)
identity = torch.eye(4)
print("zeros:")
print(zeros)
print("ones:")
print(ones)
print("random_uniform:")
print(random_uniform)
print("random_normal:")
print(random_normal)
print("identity:")
print(identity)
```

## tensor indexing

```python
x = torch.tensor([10, 20, 30, 40, 50])
print("x:", x)
print("first element:", x[0])
print("last element:", x[-1])
print("first three elements:", x[0:3])
print("from index 2 onward", x[2:])
```

## 2d tensor indexing

```python
matrix = torch.tensor([[10, 20, 30], [40, 50, 60], [70, 80, 90]])

print("martix:")
print(matrix)
print("element at row 0 column 1", matrix[0, 1])
print("first row", matrix[0])
print("second column", matrix[:, 1])
print("top left 2x2 block:")
print(matrix[0:2, 0:2])
```

## tensor reshaping

```python
x = torch.arange(12)
print("original tensor :")
print(x)
print("original tensor shape :", x.shape)

x_reshaped = x.reshape(3, 4)
print("reshaped tensor :")
print(x_reshaped)
print("reshaped tensor shape :", x_reshaped.shape)
```
