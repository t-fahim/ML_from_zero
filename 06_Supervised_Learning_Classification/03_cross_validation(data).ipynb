# Cross Validations Techniques

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import LeaveOneOut
from sklearn.model_selection import ShuffleSplit
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv("../data_set/Data_Data.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
label_encoder = LabelEncoder()
df["Purchased_encoded"] = label_encoder.fit_transform(df["Purchased"])
df = df.drop(columns=["Purchased"])
df
```

```python
X = df.drop(columns=["Purchased_encoded"])
X
```

```python
y = df["Purchased_encoded"]
y
```

```python
X = pd.get_dummies(
    X,
    columns=["Country", "Sex"],
)
X
```

```python
y.value_counts()
```

### HoldOut Validation Approach- Train And Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
model = DecisionTreeClassifier()
model.fit(X_train, y_train)
result = model.score(X_test, y_test)
print(result)
```

### K Fold Cross Validation

```python
kfold_validation = KFold(3)
model = DecisionTreeClassifier()
```

```python
result = cross_val_score(model, X, y, cv=kfold_validation)
print(result)
```

```python
print(np.mean(result))
```

### Stratified K-fold Cross Validation

```python
skfold = StratifiedKFold(n_splits=3)
model = DecisionTreeClassifier()
scores = cross_val_score(model, X, y, cv=skfold)
print(scores)
```

```python
print(np.mean(scores))
```

### Leave One Out Cross Validation(LOOCV)

```python
model = DecisionTreeClassifier()
leave_validation = LeaveOneOut()
result = cross_val_score(model, X, y, cv=leave_validation)
print(result)
```

```python
print(np.mean(result))
```

### Repeated Random Test-Train Splits
This technique is a hybrid of traditional train-test splitting and the k-fold cross-validation method. In this technique, we create random splits of the data in the training-test set manner and then repeat the process of splitting and evaluating the algorithm multiple times, just like the cross-validation method.

```python
model = DecisionTreeClassifier()
ssplit = ShuffleSplit(n_splits=10, test_size=0.3, random_state=42)
result = cross_val_score(model, X, y, cv=ssplit)
print(result)
```

```python
print(np.mean(result))
```
