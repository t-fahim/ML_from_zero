# Ensemble learning
Ensemble learning is a machine learning technique that combines multiple individual models (often called base learners or weak learners) to produce a single, stronger predictive model.

```python
import numpy as np
import pandas as pd
import xgboost as xgb
from catboost import CatBoostClassifier
import lightgbm as lgb
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import VotingClassifier
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.metrics import auc, accuracy_score, confusion_matrix, mean_squared_error
from sklearn.model_selection import cross_val_score, GridSearchCV, KFold
from sklearn.ensemble import (
    RandomForestClassifier,
    BaggingClassifier,
    AdaBoostClassifier,
    GradientBoostingClassifier,
    VotingClassifier,
)
```

```python
data = pd.read_csv("../data_set/breast-cancer.csv")
data.head()
```

```python
# Independent And dependent features
X = data.iloc[:, 2:]
y = data.iloc[:, 1]
```

## Bagging (Bootstrap Aggregating)
Bagging, short for Bootstrap Aggregating, is an ensemble machine learning technique designed to reduce variance and prevent overfitting. It works by training multiple base models (usually decision trees) in parallel on different random subsets of the training data and then combining their predictions.

```python
bg = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    max_samples=0.5,
    max_features=1.0,
    n_estimators=20,
    n_jobs=-1,  # Uses all available CPU cores
)
bg.fit(x_train, y_train)
```

```python
bg.score(x_train, y_train)
```

```python
bg.score(x_test, y_test)
```

## Boosting - Ada Boost
AdaBoost (short for Adaptive Boosting) is a sequential ensemble method designed primarily to reduce bias and turn a collection of weak learners into a single strong predictive model.

Unlike Bagging, where trees are built independently in parallel, AdaBoost builds models sequentially. Each subsequent model adapts by paying extra attention to the training samples that previous models misclassified.

```python
adb = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(max_depth=1),
    n_estimators=50,
    learning_rate=1.0,
    random_state=42,
)
adb.fit(x_train, y_train)
```

```python
adb.score(x_train, y_train)
```

```python
adb.score(x_test, y_test)
```

## GradientBoostingClassifier

Gradient Boosting is one of the most powerful and widely used machine learning algorithms for tabular data. Like AdaBoost, it builds an ensemble of trees sequentially. However, instead of adjusting sample weights, Gradient Boosting trains each new tree to predict the residuals (errors) of the previous trees using Gradient Descent.

gb = GradientBoostingClassifier(
    learning_rate=0.1,       # Shrinkage factor for each tree's contribution
    n_estimators=100,        # Number of boosting stages (trees)
    max_depth=3,             # Maximum depth of individual decision trees
    random_state=42
)
gb.fit(x_train,y_train)

## xgboost

XGBoost (Extreme Gradient Boosting) is an optimized, open-source library that implements the Gradient Boosting framework with extreme computational efficiency, built-in regularization, and advanced algorithmic tricks.


```python
# 1. Encode string labels to integers (B -> 0, M -> 1)
le = LabelEncoder()
y_train_encoded = le.fit_transform(y_train)
y_test_encoded = le.transform(y_test)
```

```python
xgb_model = xgb.XGBClassifier(
    objective="binary:logistic",
    n_estimators=1000,  # Allow model to grow up to 1000 trees
    learning_rate=0.03,  # Lower learning rate works great with early stopping
    early_stopping_rounds=10,  # Stops if val score doesn't improve for 10 trees
    eval_metric="auc",
    random_state=42,
)

# 4. Fit using the Validation set
xgb_model.fit(
    x_train,
    y_train_encoded,
    eval_set=[(x_test, y_test_encoded)],
    verbose=10,  # Prints evaluation score every 10 trees
)
```

```python
# Predict numeric classes (0 or 1)
y_pred = xgb_model.predict(x_test)

# Calculate accuracy using encoded test targets
acc = accuracy_score(y_test_encoded, y_pred)
print(f"Accuracy Score: {acc:.4f}")
```

## LightGBM

LightGBM is a fast, distributed, high performance gradient boosting framework based on decision tree algorithms, used for ranking, classification and many other machine learning tasks. It has helped Kagglers win data science competitions.

```python
# 1. Initialize with controlled complexity and early stopping
clf = lgb.LGBMClassifier(
    num_leaves=15,  # Reduced from default 31 (better for smaller datasets)
    max_depth=4,  # Restrict depth to prevent overfitting
    min_child_samples=10,  # Allow splits on smaller subsets (default is 20)
    verbosity=-1,  # Mute 'No further splits' warnings
    random_state=42,
    n_jobs=-1,  # Use all available CPU cores
)

# 2. Fit with Early Stopping using a validation set
clf.fit(
    x_train,
    y_train,
    eval_set=[(x_test, y_test)],
    eval_metric="auc",
    callbacks=[lgb.early_stopping(stopping_rounds=15, verbose=False)],
)
```

```python
preds_class = clf.predict(x_test)
```

```python
print(metrics.confusion_matrix(y_test, preds_class))
print(clf.score(x_test, y_test))
```

## CatBoost

CatBoost (short for Categorical Boosting) is a high-performance gradient boosting library developed by Yandex. It is engineered specifically to deliver state-of-the-art accuracy on tabular datasets, particularly when datasets contain categorical features (strings, categories, IDs).

While XGBoost and LightGBM often require extensive manual preprocessing (like One-Hot Encoding or Target Encoding), CatBoost handles categorical data natively under the hood with minimal setup.

## Voting Classifier - Multiple Model Ensemble 

A Voting Classifier is one of the simplest yet most effective ensemble techniques. Instead of relying on a single model or a single family of algorithms (like decision trees in Random Forests or Boosting), a Voting Classifier combines predictions from completely different machine learning algorithms (e.g., Logistic Regression, Support Vector Machines, Random Forests, XGBoost, and LightGBM).

By averaging or voting across diverse models, individual errors tend to cancel out, resulting in a more robust final prediction.

```python
lr = make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000, random_state=42))
```

```python
# Standardize feature-sensitive base estimators
lr_pipeline = make_pipeline(
    StandardScaler(), LogisticRegression(max_iter=2000, random_state=42)
)
svm_pipeline = make_pipeline(
    StandardScaler(), SVC(kernel="poly", degree=2, probability=True, random_state=42)
)
dt = DecisionTreeClassifier(max_depth=4, random_state=42)
```

```python
# 2. Build the Voting Classifier (Soft Voting recommended)
evc = VotingClassifier(
    estimators=[("logistic", lr_pipeline), ("dt", dt), ("svm", svm_pipeline)],
    voting="soft",  # Averaging predicted probabilities
)

evc.fit(x_train, y_train)
```

```python
evc.score(x_test, y_test)
```

## Stacked generalization

**Stacked generalization** consists in stacking the output of individual estimator and use a classifier to compute the final prediction. Stacking allows to use the strength of each individual estimator by using their output as input of a final estimator.


```python
from sklearn.ensemble import StackingClassifier
```

```python
stacking_clf = StackingClassifier(
    estimators=[("logistic", lr_pipeline), ("dt", dt), ("svm", svm_pipeline)],
    final_estimator=LogisticRegression(random_state=42),
    cv=5,  # Internal cross-validation folds to generate meta-features
)

stacking_clf.fit(x_train, y_train)
```

```python
y_pred = stacking_clf.predict(x_test)
```

```python
stacking_clf.score(x_test, y_test)
```
