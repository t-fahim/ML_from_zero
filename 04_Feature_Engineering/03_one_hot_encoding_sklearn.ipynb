# OneHotEncoder in Scikit-learn

`OneHotEncoder` is a preprocessing technique in **Scikit-learn** that converts categorical features into numerical binary columns (0/1). Machine learning algorithms generally cannot work directly with text categories, so OneHotEncoder transforms them into a machine-readable format.

---

## Parameters

| Parameter | Default Value | Description | Example |
|-----------|---------------|-------------|---------|
| `categories` | `'auto'` | Automatically detects unique categories from the input data. You can also specify categories manually. | `categories=[['Low', 'Medium', 'High']]` |
| `drop` | `None` | Drops one category per feature to avoid the dummy variable trap (multicollinearity). | `drop='first'`, `drop='if_binary'` |
| `sparse_output` | `True` | Returns a sparse matrix to save memory. Set to `False` to return a dense NumPy array. | `sparse_output=False` |
| `dtype` | `float` | Specifies the data type of the encoded output values. | `dtype=int`, `dtype=bool`, `dtype=float` |
| `handle_unknown` | `'error'` | Defines how to handle categories that were not seen during training when transforming new data. | `handle_unknown='ignore'` |
| `min_frequency` | `None` | Groups categories with low frequency into a single "infrequent" category. | `min_frequency=5` |
| `max_categories` | `None` | Limits the maximum number of encoded categories. Less frequent categories are grouped together. | `max_categories=10` |
| `feature_name_combiner` | `'concat'` | Controls how the output feature names are created. | `feature_name_combiner='concat'` |

---

### Explanation

- **`drop='first'`**
  - Removes the first category from each feature.

- **`handle_unknown='ignore'`**
  - Prevents errors when new unseen categories appear during prediction.
  - Encodes unknown categories as all zeros.

- **`sparse_output=False`**
  - Returns a normal NumPy array instead of a sparse matrix.

- **`dtype=int`**
  - Returns binary values as integers (`0` and `1`) instead of floating-point values (`0.0` and `1.0`).

---

# set_output(transform='pandas')

By default, `OneHotEncoder` returns a **NumPy array** after transformation.

Using:

```python
ohe = OneHotEncoder(
    sparse_output=False
).set_output(transform='pandas')
```

returns a **Pandas DataFrame** with meaningful column names.

````python
ohe = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False,
).set_output(transform="pandas")
````

````python
ohetransform_city = ohe.fit_transform(df[["city"]])
ohetransform_city
````

````python
ohetransform_size = ohe.fit_transform(df[["size"]])
ohetransform_size
````

````python
df = pd.concat([df, ohetransform_city, ohetransform_size], axis=1)
df
````

````python
df = df.drop(columns=["city", "size"])
df
````
