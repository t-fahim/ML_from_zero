# Label Encoding

**Label Encoding** is a technique that converts categorical values into numerical values by assigning a unique integer to each category.

Each unique category is mapped to a numerical code.


## Advantages

- Simple to implement.
- Keeps the dataset dimensions unchanged.
- Useful for ordinal data and tree-based models.

## Disadvantages

- Creates an artificial order between categories.
- Can reduce performance for non-ordinal data.
- Not suitable for linear models unless categories have an order.


```python
classes = ["ClassA", "ClassB", "ClassC", "ClassD"]

instances = [
    "ClassA",
    "ClassB",
    "ClassC",
    "ClassD",
    "ClassA",
    "ClassB",
    "ClassC",
    "ClassD",
    "ClassA",
    "ClassB",
]
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder = LabelEncoder()
encoded_labels = label_encoder.fit_transform(instances)

print("Encoded labels:", encoded_labels)
```

```python
original_labels = label_encoder.inverse_transform(encoded_labels)

print("Encoded labels:", encoded_labels)
print("Original labels:", original_labels)
```
