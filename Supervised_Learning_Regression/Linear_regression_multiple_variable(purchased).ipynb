```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("../data_set/purchased.csv")
df
```

```python
df.info()
```

```python
label_encoder = LabelEncoder()
df["Purchased_encoded"] = label_encoder.fit_transform(df["Purchased"])
df = df.drop(columns=["Purchased"])
df
```

```python
df["Country"].unique()
```

```python
df["Sex"].unique()
```

```python
X = df.drop(columns=["Purchased_encoded"])
X
```

```python
y = df["Purchased_encoded"]
y
```

```python
X = pd.get_dummies(
    X,
    columns=["Country", "Sex"],
)
X
```

# Perform PCA

```python
pca_full = PCA()
pca_full.fit(X)
pca_full.explained_variance_ratio_
```

```python
np.cumsum(pca_full.explained_variance_ratio_ * 100)
```

```python
pca_99 = PCA(n_components=2)
X_pca = pca_99.fit_transform(X)
X_pca
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_pca, y, test_size=0.3, random_state=42
)
print("X_train")
print(X_train.shape)
print("X_test")
print(X_test.shape)
print("y_train")
print(y_train.shape)
print("y_test")
print(y_test.shape)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)

print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)
```

```python
predictions = model.predict(X_test)

print(predictions)
```
